```python
def f(x):
    return (x+3)**2
```

```python
def derivative_f(x):
    return 2 * (x+3)
```

```python
x = 2
learning_rate = 0.1
iterations = 100
```

```python
for i in range(iterations):
    x = x - learning_rate * derivative_f(x)
```

```python
print("Local minimum point x = ", round(x, 6))
print("Minimum value y = ", round(f(x), 6))
```
